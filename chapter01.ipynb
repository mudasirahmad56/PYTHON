### operators in python

```python
# arithmetic operations
# Relational operators
# Logical operators
# Bitwise operators
# assignment operators
# membership operators
```

```python
# Arithmetic operators
print(5 + 2)    # addition
print(5 - 2)    # subtraction
print(5 * 2)    # multiplication
print(5 / 2)    # division
print(5 // 2)   # floor division
print(5 % 2)    # modulus
print(5 ** 2)   # exponentiation
```

```python
# Relational operators
print(5 > 2)    # greater than
print(5 < 2)    # less than
print(5 >= 5)   # greater than or equal to
print(5 <= 5)   # less than or equal to
print(5 == 5)   # equal to
print(5 != 5)   # not equal to
```

```python
# Logical operators
print(1 and 0)    # logical AND
print(1 or 0)     # logical OR
print(not 1)      # logical NOT
```

```python
# bitwise operators
print(5 & 3)    # bitwise AND
print(5 | 3)    # bitwise OR
print(5 ^ 3)    # bitwise XOR
print(~5)       # bitwise NOT
print(5 << 1)  # left shift
print(5 >> 1)  # right shift
```

### *args and **kwargs
#### *args and **kwargs
are special python keywords that are used to pass to variable length of arguments to a function

```python
# *agrs
def multiply(*args):
    product = 1
    for i in args:
        product = product * i
    return product
```

```python
multiply(1,2,3,4,2,34,344,34,43,4,3555,23)
```

```python
#**kwargs
def display(**kwargs):
    for (key, value) in kwargs.items():
        print(key,'->',value)
```

```python
display(pakistan = 'islamabad', india = 'delhi', srilanka = 'colombo')
```

#### instance veriable

```python
class student:
    def __init__(self,name,age,grade):
        self.name = name
        self.age = age
        self.grade = grade
    def display_info(self):
        print(f"student Name:{self.name}")
        print(f"student Age: {self.age}")
        print(f"student Grade: {self.grade}")

    def is_eligible(self):
        if self.age >= 18:
            print(self.name,"is eligible for admission)")
        else:
            print(self.name,"is not eligible for admission")
```
